```python
from rapidfuzz.fuzz import ratio

companies = (
    df["Entreprise"]
    .dropna()
    .drop_duplicates()
    .tolist()
)

similar_pairs = []

for i in range(len(companies)):
    for j in range(i + 1, len(companies)):
        score = ratio(companies[i], companies[j])

        if score >= 85:
            similar_pairs.append({
                "Entreprise 1": companies[i],
                "Entreprise 2": companies[j],
                "similarité": round(score, 2)
            })

similar_pairs = pd.DataFrame(similar_pairs)

display(
    similar_pairs
    .sort_values("similarité", ascending=False)
)
```

```python
import re

def normalize_company(name):
    if pd.isna(name):
        return None

    name = str(name).strip().upper()

    # Accents
    name = unicodedata.normalize("NFD", name)
    name = "".join(
        c for c in name
        if unicodedata.category(c) != "Mn"
    )

    # Apostrophes typographiques
    name = name.replace("’", "'")

    # S.A., S.A → SA
    name = re.sub(r"\bS\.A\.?\b", "SA", name)

    # Ponctuation autour des espaces
    name = re.sub(r"\s*-\s*", " ", name)

    # Espaces multiples
    name = re.sub(r"\s+", " ", name)

    return name.strip()
```

```python
# Garder les paires avec une similarité >= 95 %
high_similarity = similar_pairs[
    similar_pairs["similarité"] >= 95
].copy()

# ============================================================
# Regroupement des variantes
# ============================================================

parent = {}

def find(x):
    if parent[x] != x:
        parent[x] = find(parent[x])
    return parent[x]

def union(a, b):
    if a not in parent:
        parent[a] = a
    if b not in parent:
        parent[b] = b

    root_a = find(a)
    root_b = find(b)

    if root_a != root_b:
        parent[root_b] = root_a


# Construire les groupes de similarité
for _, row in high_similarity.iterrows():
    union(row["Entreprise 1"], row["Entreprise 2"])


# Créer le mapping
groups = {}

for company in parent:
    root = find(company)
    groups.setdefault(root, []).append(company)

company_mapping = {}

for root, companies in groups.items():
    reference = companies[0]

    for company in companies:
        company_mapping[company] = reference


# Appliquer la fusion directement dans la colonne Entreprise
df["Entreprise"] = df["Entreprise"].replace(company_mapping)

print("Nombre d'entreprises :", df["Entreprise"].nunique())
```

```python
# ============================================================
# FUSION DES ENTREPRISES SIMILAIRES ENTRE 90 % ET 95 %
# ============================================================

# Paires entre 90 % et 95 %
similar_90_95 = similar_pairs[
    (similar_pairs["similarité"] >= 90) &
    (similar_pairs["similarité"] < 95)
].copy()

# Cas à NE PAS fusionner
excluded_pairs = {
    (
        "THALES LAS FRANCE SAS",
        "THALES DMS FRANCE SAS"
    ),
    (
        "MANUFACTURE D'HORLOGERIE AUDEMARS PIGUET & CIE",
        "MANUFACTURE D'HORLOGERIE AUDEMARS PIGUET SA"
    )
}

# ============================================================
# Union-Find
# ============================================================

parent = {}

def find(x):
    if parent[x] != x:
        parent[x] = find(parent[x])
    return parent[x]

def union(a, b):
    if a not in parent:
        parent[a] = a
    if b not in parent:
        parent[b] = b

    root_a = find(a)
    root_b = find(b)

    if root_a != root_b:
        parent[root_b] = root_a


# Construire les groupes
for _, row in similar_90_95.iterrows():

    pair = (
        row["Entreprise 1"],
        row["Entreprise 2"]
    )

    # Vérifier aussi l'ordre inverse
    reverse_pair = (
        row["Entreprise 2"],
        row["Entreprise 1"]
    )

    if pair in excluded_pairs or reverse_pair in excluded_pairs:
        continue

    union(
        row["Entreprise 1"],
        row["Entreprise 2"]
    )


# ============================================================
# Créer le mapping
# ============================================================

groups = {}

for company in parent:
    root = find(company)
    groups.setdefault(root, []).append(company)

company_mapping = {}

for root, companies in groups.items():
    reference = companies[0]

    for company in companies:
        company_mapping[company] = reference


# ============================================================
# Appliquer la fusion
# ============================================================

df["Entreprise"] = df["Entreprise"].replace(company_mapping)

print("Nombre d'entreprises :", df["Entreprise"].nunique())
```

```python
import unicodedata
import re

def normalize_column_name(col):
    col = str(col).strip().lower()

    # Supprimer les accents
    col = unicodedata.normalize("NFD", col)
    col = "".join(
        c for c in col
        if unicodedata.category(c) != "Mn"
    )

    # Remplacer les caractères non alphanumériques par _
    col = re.sub(r"[^a-z0-9]+", "_", col)

    # Supprimer les _ au début/à la fin
    col = col.strip("_")

    return col

df.columns = [normalize_column_name(col) for col in df.columns]

print(df.columns.tolist())
```

```python
# ============================================================
# NORMALISATION — PAYS
# ============================================================

import unicodedata

def normalize_text(value):
    if pd.isna(value):
        return value

    value = str(value).strip().upper()

    value = unicodedata.normalize("NFD", value)
    value = "".join(
        c for c in value
        if unicodedata.category(c) != "Mn"
    )

    value = " ".join(value.split())

    return value


df["pays_lieu_stage"] = df["pays_lieu_stage"].map(normalize_text)

print("Valeurs distinctes après normalisation :")
print(
    df["pays_lieu_stage"]
    .value_counts(dropna=False)
    .sort_index()
    .to_string()
)
```

```python
# ============================================================
# NORMALISATION — VILLES
# ============================================================

def normalize_city(value):
    if pd.isna(value):
        return value

    value = str(value).strip().upper()

    # Suppression des accents
    value = unicodedata.normalize("NFD", value)
    value = "".join(
        c for c in value
        if unicodedata.category(c) != "Mn"
    )

    # Normalisation des espaces
    value = " ".join(value.split())

    # Normalisation des espaces autour des tirets
    value = re.sub(r"\s*-\s*", "-", value)

    return value


df["ville"] = df["ville"].map(normalize_city)
df["ville_lieu_stage"] = df["ville_lieu_stage"].map(normalize_city)

print("Nombre de villes distinctes après normalisation :")
print(f"ville              : {df['ville'].nunique(dropna=True)}")
print(f"ville_lieu_stage   : {df['ville_lieu_stage'].nunique(dropna=True)}")
```

```python
# ============================================================
# NORMALISATION SUPPLÉMENTAIRE — VILLES
# ============================================================

def normalize_city_for_search(value):
    if pd.isna(value):
        return value

    value = str(value).strip().upper()

    # Suppression des accents
    value = unicodedata.normalize("NFD", value)
    value = "".join(
        c for c in value
        if unicodedata.category(c) != "Mn"
    )

    # Tirets et espaces autour des tirets → espace
    value = re.sub(r"\s*-\s*", " ", value)

    # Espaces multiples
    value = " ".join(value.split())

    return value


df["ville"] = df["ville"].map(normalize_city_for_search)
df["ville_lieu_stage"] = df["ville_lieu_stage"].map(normalize_city_for_search)
```

```python
# ============================================================
# NORMALISATION — ENTREPRISES
# ============================================================

def normalize_company_for_search(value):
    if pd.isna(value):
        return value

    value = str(value).strip().upper()

    # Suppression des accents
    value = unicodedata.normalize("NFD", value)
    value = "".join(
        c for c in value
        if unicodedata.category(c) != "Mn"
    )

    # Espaces multiples
    value = " ".join(value.split())

    return value


df["entreprise"] = df["entreprise"].map(normalize_company_for_search)

print(f"Nombre d'entreprises distinctes : {df['entreprise'].nunique(dropna=True)}")
print(f"Valeurs manquantes : {df['entreprise'].isna().sum()}")
```
