```python
from rapidfuzz.fuzz import ratio

companies = (
    df["Entreprise"]
    .dropna()
    .drop_duplicates()
    .tolist()
)

similar_pairs = []

for i in range(len(companies)):
    for j in range(i + 1, len(companies)):
        score = ratio(companies[i], companies[j])

        if score >= 85:
            similar_pairs.append({
                "Entreprise 1": companies[i],
                "Entreprise 2": companies[j],
                "similarité": round(score, 2)
            })

similar_pairs = pd.DataFrame(similar_pairs)

display(
    similar_pairs
    .sort_values("similarité", ascending=False)
)
```

```python
import re

def normalize_company(name):
    if pd.isna(name):
        return None

    name = str(name).strip().upper()

    # Accents
    name = unicodedata.normalize("NFD", name)
    name = "".join(
        c for c in name
        if unicodedata.category(c) != "Mn"
    )

    # Apostrophes typographiques
    name = name.replace("’", "'")

    # S.A., S.A → SA
    name = re.sub(r"\bS\.A\.?\b", "SA", name)

    # Ponctuation autour des espaces
    name = re.sub(r"\s*-\s*", " ", name)

    # Espaces multiples
    name = re.sub(r"\s+", " ", name)

    return name.strip()
```

```python
# Garder les paires avec une similarité >= 95 %
high_similarity = similar_pairs[
    similar_pairs["similarité"] >= 95
].copy()

# ============================================================
# Regroupement des variantes
# ============================================================

parent = {}

def find(x):
    if parent[x] != x:
        parent[x] = find(parent[x])
    return parent[x]

def union(a, b):
    if a not in parent:
        parent[a] = a
    if b not in parent:
        parent[b] = b

    root_a = find(a)
    root_b = find(b)

    if root_a != root_b:
        parent[root_b] = root_a


# Construire les groupes de similarité
for _, row in high_similarity.iterrows():
    union(row["Entreprise 1"], row["Entreprise 2"])


# Créer le mapping
groups = {}

for company in parent:
    root = find(company)
    groups.setdefault(root, []).append(company)

company_mapping = {}

for root, companies in groups.items():
    reference = companies[0]

    for company in companies:
        company_mapping[company] = reference


# Appliquer la fusion directement dans la colonne Entreprise
df["Entreprise"] = df["Entreprise"].replace(company_mapping)

print("Nombre d'entreprises :", df["Entreprise"].nunique())
```

```python
# ============================================================
# FUSION DES ENTREPRISES SIMILAIRES ENTRE 90 % ET 95 %
# ============================================================

# Paires entre 90 % et 95 %
similar_90_95 = similar_pairs[
    (similar_pairs["similarité"] >= 90) &
    (similar_pairs["similarité"] < 95)
].copy()

# Cas à NE PAS fusionner
excluded_pairs = {
    (
        "THALES LAS FRANCE SAS",
        "THALES DMS FRANCE SAS"
    ),
    (
        "MANUFACTURE D'HORLOGERIE AUDEMARS PIGUET & CIE",
        "MANUFACTURE D'HORLOGERIE AUDEMARS PIGUET SA"
    )
}

# ============================================================
# Union-Find
# ============================================================

parent = {}

def find(x):
    if parent[x] != x:
        parent[x] = find(parent[x])
    return parent[x]

def union(a, b):
    if a not in parent:
        parent[a] = a
    if b not in parent:
        parent[b] = b

    root_a = find(a)
    root_b = find(b)

    if root_a != root_b:
        parent[root_b] = root_a


# Construire les groupes
for _, row in similar_90_95.iterrows():

    pair = (
        row["Entreprise 1"],
        row["Entreprise 2"]
    )

    # Vérifier aussi l'ordre inverse
    reverse_pair = (
        row["Entreprise 2"],
        row["Entreprise 1"]
    )

    if pair in excluded_pairs or reverse_pair in excluded_pairs:
        continue

    union(
        row["Entreprise 1"],
        row["Entreprise 2"]
    )


# ============================================================
# Créer le mapping
# ============================================================

groups = {}

for company in parent:
    root = find(company)
    groups.setdefault(root, []).append(company)

company_mapping = {}

for root, companies in groups.items():
    reference = companies[0]

    for company in companies:
        company_mapping[company] = reference


# ============================================================
# Appliquer la fusion
# ============================================================

df["Entreprise"] = df["Entreprise"].replace(company_mapping)

print("Nombre d'entreprises :", df["Entreprise"].nunique())
```
